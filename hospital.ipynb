```python
import pandas as pd
data=pd.read_csv('hospital data.csv')
data
```

```python
#find errors and duplicates with null
print(data.shape)
print(data.isnull().sum())
print(data.duplicated().sum())
print(data.loc[data.duplicated().sum()])
```

```python
#remove duplicate::

data=data.drop_duplicates()
data
```

```python
data=data.dropna()
data
```

```python
t=data.isnull().sum()
t
```

```python
t=data.dtypes
t
```

```python
data['admission_date']=pd.to_datetime(data['admission_date'])
data['discharge_date']=pd.to_datetime(data['discharge_date'])
print(data.dtypes)
```

```python
f=data['insurance_covered_usd']-data['treatment_cost_usd']
data['amount_paid']=abs(f)
data
```

```python
def add(amt):
    if amt>=10000:
        data['status']='high paid'
    elif amt>=5000:
        data['status']='medium paid'
    else:
        data['status']='low paid'
data=data['amount_paid'].apply(add)
data
```

```python
g=data.groupby('department')['amount_paid'].agg(['sum','count','max','min','mean'])
g
```

```python
h=pd.pivot_table(
    data=data,
    index='department',
    columns='admission_type',
    values='amount_paid',
    aggfunc='count',
    fill_value=0,
    margins=True,
    margins_name='no of count in department'
)
h
```
