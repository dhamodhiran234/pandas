```python
import pandas as pd
data=pd.read_csv('pd.csv')
data
```

```python
#find error::

print(data.shape)
print(data.duplicated().sum())
print(data.isnull().sum())
print(data.loc[data.duplicated()])
```

```python
#data cleaning::

data=data.drop_duplicates()
data
```

```python
print(data.shape)
```

```python
print(data.dtypes)
```

```python
data['discount']=data['discount'].fillna(0)
data
```

```python
print(data.isnull().sum())
```

```python
data['payment_method']=data['payment_method'].fillna('unknown')
data['customer_rating']=data['customer_rating'].fillna(data['customer_rating'].median())
data
```

```python
print(data.isnull().sum())
```

```python
# column datatype conversions::
data['signup_date']=pd.to_datetime(data['signup_date'])
data
```

```python
data['transaction_date']=pd.to_datetime(data['transaction_date'])
print(data.dtypes)
```

```python
data['price']=data['price'].astype(float)
print(data.dtypes)
```

```python
data['gross_amount']=data['price']*data['quantity']
data['discount_amount']=data['gross_amount']*data['discount']
data['final_amount']=data['gross_amount']-data['discount_amount']
data
```
