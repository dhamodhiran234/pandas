```python
import pandas as pd
data=pd.read_csv('pd2.csv')
data
```

```python
#find error::

print(data.shape)
print(data.duplicated().sum())
print(data.loc[data.duplicated()])
print(data.isnull().sum())
```

```python
#handle duplicates and error::

data=data.drop_duplicates()
print(data.duplicated().sum())
```

```python
print(data.dtypes)
```

```python
data['performance_score']=data['performance_score'].fillna(0)
data['education_level']=data['education_level'].fillna('unknown')
print(data.isnull().sum())
```

```python
print(data.dtypes)
```

```python
data['hire_date']=pd.to_datetime(data['hire_date'])
print(data.dtypes)
```

```python
def bonus(amt):
    if amt>10:
        return 'good'
    else:
        return 'bad'
data['status']=data['projects_completed'].apply(bonus)
data
```

```python

def div(row):
    mul=row['projects_completed']*row['performance_score']
    return row['salary_usd'] -mul
    
data['per']=data.apply(div,axis=1)
data
```

```python

p=pd.pivot_table(
    data=data,
    index='department',
    columns='job_title',
    values='salary_usd',
    aggfunc=['sum','count'],
    fill_value=-0,
    margins=True,
    margins_name='total final'
    
)
p


```

```python
mi=data.head(5)
print(mi)
import matplotlib.pyplot as plt
plt.scatter(mi['performance_score'],mi['salary_usd'],marker='+',color='red')
plt.plot(mi['performance_score'],mi['salary_usd'])
plt.show()
```

```python
from sklearn import linear_model
reg=linear_model.LinearRegression()
x=mi[['performance_score']]
y=mi['salary_usd']
reg.fit(x,y)
y_pred=reg.predict(mi[['performance_score']])
y_pred
```

```python
plt.scatter(mi['performance_score'],mi['salary_usd'],marker='+',color='red')
plt.plot(mi['performance_score'],y_pred)
plt.show()
```

```python
error=sum((y-y_pred)**2)/5
error
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
msa=mean_squared_error(y,y_pred)
msa
from sklearn.metrics import mean_squared_error,mean_absolute_error

mae=mean_absolute_error(y,y_pred)
mae
```
