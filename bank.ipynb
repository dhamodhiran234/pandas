```python
# clean null,duplicate
data['transaction_type']=data['transaction_type'].fillna('unknown')
data['merchant_category']=data['merchant_category'].fillna('unknown')
print(data.isnull().sum())
data=data.drop_duplicates()
data
```

```python
data['balance_before_txn_usd']=data['balance_after_txn_usd']-data['amount_usd']
data
```

```python
t=data.groupby('transaction_type')[['balance_after_txn_usd','balance_before_txn_usd']].agg(['sum','count','min','max','mean'])
t
```

```python
t=data['balance_after_txn_usd'].agg(['sum','count'])
t
```

```python
t=pd.pivot_table(
    data=data,
    index='transaction_type',
    columns='channel',
    values='balance_after_txn_usd',
    aggfunc=['sum','count'],
    margins=True,
    fill_value=0,
    margins_name='total amount'
)
t
```
